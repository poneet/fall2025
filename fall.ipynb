```python
# Phys4030 code challenge email needs to contain elements of the diffusion constant derivation...
import matplotlib.pyplot as plt
import numpy as np

# ==================================================================================================================================
#       .__   .__  .___  .  .  . .   , . ____  .   .    .__ __  .___ ____  . .___  .   .   // _.-''''-._
#      /__.' /__/ /   / /  /  / /\  / / /___/ /\  /    /  /  / /   /  /   / /   / /\  /      /   _.._   \
#     /___/ /  \ /___/ /__/__/ /  \/ / /   / /  \/    /     / /___/  /   / /___/ /  \/      |   (    )   |
#                                                                                            \   '--'   / 
#                                                                                             '-..__..-' //       
#
def randomWalk(numOfWalkers,numOfSteps):
    # Simulates a random walk (diffusion) in one dimension and plots a pdf, entropy vs time and the results of the random walk
    # Parameter:      numOfWalkers       - The number of random walkers.
    # Parameter:      numOfSteps         - The number of steps taken, at 1 step/second also represents time in seconds.
    # Returns  :      nothing
    # Local Variable: DIRECTIONS         - Random numbers between 0 and 1 representing steps.
    # Local Variable: N                  - Reassignment of numOfSteps.
    # Local Variable: s                  - Modified DIRECTIONS to contain only -1 or 1, representing left and right steps respectively for each walker.
    # Local Variable: x                  - X-displacements for each walker assigned using np.sum()
    N = numOfSteps
    s = np.zeros( (numOfWalkers,N) )
    walker = 0# Walker number
    while(walker<numOfWalkers):
        DIRECTIONS = np.random.randint(0,2,N)# 0=Left, 1=Right 
        i = 0# go through and set all DIRECTIONS[i] equal to 0's to -1
        while (i<N):
            if (DIRECTIONS[i] == 0):# There should be no stationary steps, so 0=Left
                DIRECTIONS[i] = -1# Left
            i += 1
        s[walker] = DIRECTIONS# These are the steps each walker took
        walker += 1
    # ===============================================================================================
    #
    # Entropy S = - Σ p_i * ln(p_i) 
    #
    # ===============================================================================================
    Entropy = np.zeros(N)
    t = 0
    while(t<N):
        L = 1# Step-size, one step per time interval
        walker = 0# Walker number
        x = np.zeros(numOfWalkers)# This will contain the total x displacement each walker took at time t, for numOfWalkers walkers
        while (walker<numOfWalkers):
            # print("----- DEBUG from randomWalk() ----- use a single walker and debug, this is the steps taken by s[walker][0:t+1] = ", s[walker][0:t+1])
            x[walker] = np.sum( s[walker][0:t+1] )*L# Total summed x displacement each walker took at time t [:t], one step per second upto N steps in array "s"
            walker += 1
            
        # --------------------------------------------------
        # Giodano - Computational Physics ch.9.7
        # --------------------------------------------------
        # Create the probabilty density function (pdf) in Fig.7.17 to calculate entropy for each time t
        i = 0
        pdf = np.zeros(N*2+1)# pdf has a negative and positive tail
        center = int( N )# center the pdf on zero
        walker=0# Walker number
        while(walker<numOfWalkers):
            pdf[center+int(x[walker])] += 1# The total x displacement for each walker is plotted on the pdf axis
            walker+=1
        pdf = pdf / numOfWalkers# Normalize to obtain probability

        # Calculate Entropy at time t
        # S = - Σ p_i * ln(p_i)
        S = 0
        i = 0
        while(i<len(pdf)):
            p_i = pdf[i]
            if (p_i > 0):
                S += - p_i * np.log(p_i)
            i+=1
        Entropy[t] = S
        t+=1# increase time, 1 step per second upto N steps, so N is also time in seconds
    # ===============================================================================================
    # 
    # Plots
    #
    # ===============================================================================================
    figwidth = 20
    figheight = 5
    # --------------------------------------------------
    # Plot: One pdf (used for Entropy calculations)
    fig1,ax1 = plt.subplots()
    ax1.plot(np.linspace(-N,N,2*N+1),pdf)
    xlabel1 = "x"
    ylabel1 = "Probability"
    title1 = "Probability Density Function for a Random Walk in One Dimension \nNumber of Steps = "+str(N)
    ax1.set_xlabel(xlabel1)
    ax1.set_ylabel(ylabel1)
    ax1.set_title(title1)
    fig1.set_figheight(figheight)
    fig1.set_figwidth(figwidth)
    # --------------------------------------------------
    # Plot: Entropy
    fig,ax = plt.subplots()
    ax.plot(np.linspace(0,N,N),Entropy)
    xlabel = "Time (s)"
    ylabel = "Entropy"
    title = "Entropy versus Time of a Random Walk in One Dimension \nNumber of Steps = "+str(N)+"; 1 step/second"
    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    ax.set_title(title)
    fig.set_figheight(figheight)
    fig.set_figwidth(figwidth)
    # --------------------------------------------------
    # Plot: The Random Walkers
    walker = 0
    y = np.zeros(numOfWalkers)# This will contain the sum of all steps
    while (walker<numOfWalkers):
        y[walker] = np.sum(s[walker])*L# The sum of all steps each walker took
        walker += 1
    fig2, ax2 = plt.subplots()
    X = np.linspace(1,numOfWalkers,numOfWalkers)
    ax2.plot([0,0],[0,numOfWalkers],'c') # vertical line at 0
    ax2.scatter(y,X,c='#ffffff',linewidths=3,edgecolors='#222222',s=50) # cells/pollen
    ax2.scatter(np.abs( np.sort(y) ),X,marker='x',color='c')# This is similar to the pdf
    ax2.set_facecolor('#777777')
    ax2.set_xlabel("Displacement")
    ax2.set_ylabel("Walker")
    fig2.set_figheight(figheight)
    fig2.set_figwidth(figwidth)
    # --------------------------------------------------
    return()

# Parameters
N = 500
numOfWalkers = 500# try decreasing this to one
randomWalk(numOfWalkers,N)

plt.show()
```
